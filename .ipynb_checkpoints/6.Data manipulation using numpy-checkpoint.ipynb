```python
import numpy as np
```

```python
var_1=np.array([1,2,3,4])
print(var_1)
```

```python
var_2=np.array([[1,2,3],[4,5,6]])
print(var_2)
```

```python
#array of zero:-
var_3=np.zeros((3,3),dtype=int)
print(var_3)
```

```python
#array of ones:-
var_4=np.ones((4,4),dtype=int)
print(var_4)
```

```python
#array of random elements:-
#here we can not change the data type as it gives random value.
#everytime we run it it gives differeent values.
var_5=np.random.rand(4,4)
print(var_5)
```

```python
#fixing the values of random arrays.
#you can give any number it dosent have any association with the arrays .
np.random.seed(4)
var_5=np.random.rand(4,4)
print(var_5)
```

```python
# here n in randn means negative so it will also show negative values.
#the values can be fixed by the same method as above.
np.random.seed(1)
var_6=np.random.randn(4,4)
print(var_6)
```

```python
# here randint is used to get only integer values . 
#and we have to mention the range
#same method will be used to fix the numbers.
np.random.seed(1)
var_7=np.random.randint(low=50,high=100,size=(4,4))
print(var_7)
```

__SLICING AND INDEXING:-
    ->INEXING:- (row ,coulmn)
        INEXING ACROSS THE ROWS 
        INEXING ACROSS THE COLUMNS
    ->SLICING:- x(for row starting index position: ending position,column same as row)
        WE WILL PERFORM SLICING IN SAME MANNER LIKE WE USE TO O IN LIST AND TUPE.
        HERE WE WILL EXTRACT A SUB MATRIX FROM A COMPLETE MATRIX.

```python
#INEXING
print(var_7)
```

```python
print(var_7[1,1])
```

```python
#SLICING:
print(var_7[0:2,0:2])
```

```python
print(var_7[2:3,2:])
```

```python
#CREATION OF IDENTITY MATRIX:
np.identity(3,dtype=int)
```

```python
#RESHAPE:
a=np.arange(1,13).reshape(3,4)
print(a)
```

```python
#CONVERT 2D INTO 1D:
a=a.ravel()
print(a)
```

```python
#TRANSPOSE:
b=np.arange(1,13).reshape(4,3)
print(b)
```

```python
b.transpose()
```

```python
c=np.arange(12).reshape(3,4)
d=np.arange(12,24).reshape(3,4)
```

```python
print(c)
```

```python
print(d)
```

```python
#STACKING: 
#--here v refers to vertival stacking.
e=np.vstack((c,d))
print(e)
```

```python
#--here h refers to horizontal stacking.
f=np.hstack((c,d))
print(f)
```

```python
#SPLITTING:-
#here h is horizontal splitting
np.hsplit(e,2)
```

```python
#here v is vertically splitting.
np.vsplit(e,3)
```
