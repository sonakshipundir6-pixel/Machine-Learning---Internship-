```python
import numpy as np
```

```python
array_1=np.array([1,2,3])  #it will by default look like list.
print(array_1)
```

```python
array_2=np.array((38,40,50))
print(array_2)
```

```python
array_3=np.array([[1,2,3],[4,5,6]])
print(array_3)
```

```python
array_4=np.array([[1,2,3],[4,5,6],[5,6,7]])
print(array_4)
```

```python
# BASIC METHODS THAT CAN BE USED IN NUMPY.
np.array([1,2,3,4],dtype="complex")
```

```python
np.array([1,2,3,4],dtype="float")
```

```python
print(array_1+array_2)
```

```python
list_1=[1,2,3]
list_2=[30,40,50]
print(list_1+list_2)  #here it do concatination
```

```python
list_ans=[]
for i, j in zip(list_1,list_2):
    list_ans.append(i+j)
print(list_ans)
```

```python
a1=np.arange(10)    #it only creates 1d array.and helps to create array same as range function.
print(a1)
```

```python
a2=np.arange(12,dtype="float")
print(a2)
```

```python
a2=np.arange(12,dtype="float").reshape(4,3) #dtype changes data type.
print(a2)
```

```python
a1=np.arange(10).reshape(5,2)  #here 1d is converted to 2d.
print(a1)
```

```python
a2=a2.astype(np.int64)    #changing the data tupe of the array.
print(a2)
```

```python
# basic operations on array.
a2*4
```

```python
#running for loop on array. it also picks rows.
for i in a2:
    print(i)
```

```python
#when you want to reach out the elements or values of array or matrix.
for i in np.nditer(a2):
    print(i)
```

```python
print(a2>5)
```
