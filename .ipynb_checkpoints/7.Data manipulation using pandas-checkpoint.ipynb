```python
# Checking of duplicate values:- everything will be copy of one column.

df.duplicated()
```

```python
df.duplicated().sum() 
#gives sum of all the values.
```

```python
#What if it has duplicate set of values? how to remove them??

df=df.drop_duplicates()
print(df)
```

```python
#Null Values:- 
df.info()
# it will give tell whether there are any duplicate values or not.
#if we have null values we need to keep them for imputation.
# Imputation - it means filling the null values with statistical view of data set.
#statistical overview of the catagorical type is different from that of numerical type.
# that means we cant perfrom same operations on them.
# for catagorical nature we will find the mode of the data set.
#for numerical nature we will find the median of the data set.
```

```python
df.isnull().sum()
```

```python
#finding mode :-
#[0] will give only the value.
df["loan_limit"].mode()[0]
```

```python
#for catagorical:-

df["loan_limit"]=df["loan_limit"].fillna(df["loan_limit"].mode()[0])
```

```python
#for numerical:-
df["LTV"]=df["LTV"].fillna(df["LTV"].median)
```

```python
df
```

```python
#creating feature to create ratio between 2 columns.

df["ratio"]=df["Credit_Score"]/df["dtir1"]
```

```python
df
```

```python
df["ratio"].describe()
```

```python
#binning- segregate data set based on different ranges or criteria.
#if person has gender as male , loan purpose as p1 and credit worthiness --> stable 
#else: unstable

def bin(x):
    if (x["Gender"]=="Male") & (x["loan_purpose"]=="p1") & (x["Credit_Worthiness"]=="l1"):
        return"Stable"
    else:
        return"Unstable"
```

```python
# axis=1 is for column and axis=0 is for rows.
df["Bin"]=df.apply(bin,axis=1)
```
