```python
import numpy as np
import matplotlib.pyplot as plt
```

```python

marks= np.array([72, 85, 90, 65, 78, 92, 68, 75, 88, 80])

print("Mean:",np.mean(marks))
print("Median:",np.median(marks))
print("Range:",np.max(marks)-np.min(marks))
print("Sample Variance:",np.var(marks,ddof=1))
print("Std Deviation (sample):",np.std(marks, ddof=1))
```

```python
import pandas as pd

df=pd.read_csv("C:/Users/Asus/employees.csv")
```

```python
average_salary= df.groupby("department")["salary"].mean()
print(average_salary)
```

```python
experienced_employees= df[df["experience_yrs"]>5]
print(experienced_employees)
```

```python
top_3= df.sort_values("salary",ascending=False).head(3)
print(top_3)
```

```python
df["salary_per_year_exp"]= df["salary"]/df["experience_yrs"]
print(df.round(2))
```

```python
plt.figure(figsize=(8, 5))
plt.bar(average_salary.index, average_salary.values)

plt.title("Average Salary by Department")
plt.xlabel("Department")
plt.ylabel("Average Salary")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

for department in df["department"].unique():
    department_data = df[df["department"] == department]
    plt.scatter(
        department_data["experience_yrs"],
        department_data["salary"],
        label=department
    )

plt.title("Experience vs Salary by Department")
plt.xlabel("Experience (Years)")
plt.ylabel("Salary")
plt.legend()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.hist(df["age"], bins=5, edgecolor="black")

plt.title("Distribution of Employee Ages")
plt.xlabel("Age")
plt.ylabel("Number of Employees")
plt.show()
```
