```python
import pandas as pd
```

```python
# HOW TO CREATE SERIES???
S1= pd.Series(data=[10,20,30,40],index=['Maths','English','Hindi','Science'])
print(S1)
```

```python
S2= pd.Series(data=[13,24,33,49],index=['Maths','English','Hindi','Science'])
print(S2)
```

```python
#CREATING A DATA FRAME: key will become the column name.
#1st way:
df_1=pd.DataFrame(data={'English':[10,20,30,40],'Hindi':[10,20,30,67]},index=['a','b','c','d'])
print(df_1)
```

```python
#2nd way:
pd.DataFrame(data={'col_1':S1,'col_2':S2}) #series should have same number of elements.
```
