```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
#LINE CHART:-
x=[1,2,3,4,7]
y=[10,2,5,9,6]

plt.plot(x,y)
```

```python
help(plt.plot)
#gives all the parameters that can be used.
```

```python
plt.plot(x,y,marker='o')
```

```python
plt.plot(x,y,marker='o',color='c')
```

```python
# BAR PLOT: to distinguish different types of values.
x_1=[5,2,9,4,7]
y_1=[10,20,34,23,21]
plt.bar(x_1,y_1)
```

```python
help(plt.bar)
```

```python
plt.bar(x_1,y_1,color=['pink','green','blue','red','brown'])
```

```python
plt.bar(x_1,y_1,color=['pink','green','blue','red','brown'])
plt.xlabel('Sales')
plt.ylabel('Profit')
plt.title('Sales Portfolio')
```

```python
# HISTOGRAM : used for numerical type of continuous data.  it only has one axis.
# purpose: it is used to perform data analysis in machine learning.
axis_1=[10,20,3,56]
plt.hist(y)
```

```python
help(plt.hist)
```

```python
# SCATTER GRAPH: it just shows dots.
x=[10,20,30,40]
y=[10,5,8,4]
plt.scatter(x,y,color='r')
```

```python
help(plt.scatter)
```

```python
#STACKED PLOT : one graph upon another.
var_1=['A','B','C','D']
y1=[10,20,10,45]
y2=[20,25,15,85]
plt.bar(var_1,y2,color='red')
plt.bar(var_1,y1)
```

```python
# PIE CHART: used for differencing.
y_n=np.array([35,25,25,15])
label_1=['Men','Women','Others','Unidentified']
plt.pie(y_n,labels=label_1)
```

```python
help(plt.pie)
```

```python
plt.pie(y_n,labels=label_1,explode=[0.1,0.2,0.3,0])
plt.legend() #give a map of the data .
```
