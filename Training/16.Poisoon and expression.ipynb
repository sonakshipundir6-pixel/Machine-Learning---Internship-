```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import poisson
```

```python
#parameters :
lamda_value=5  #average rate of constant mean.
k=np.arange(0,15)  #possible valus for k or t.


#poisson
#pmf= probability mass function

pmf_value=poisson.pmf(k,lamda_value)

#plotting;

plt.bar(k,pmf_value,color="red",alpha=0.9)
#alpha is used for brightness of color.

plt.title("Poisson Distribution")
plt.xlabel("Number of Events")
plt.ylabel("Probability")
plt.grid(axis="y")
```

```python
from scipy.stats import expon

lambda_value1=1

#generate the value of x:
x=np.linspace(0,10,100)
#linspace creates an array with floating values.

pdf_value=expon.pdf(x,scale=1/lambda_value1)

#plotting:
plt.plot(x,pdf_value,color="red",alpha=0.9)
plt.title("Exponential Distribution")
plt.xlabel("Sample")
plt.ylabel("pdf")
plt.grid()
```
