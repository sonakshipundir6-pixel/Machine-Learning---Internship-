```python
import pandas as pd
import seaborn as sns
from sklearn.model_selection import train_test_split
```

```python
df=pd.read_csv("C:/Users/Asus/Downloads/Company_Profit_lyst1729240579399.csv")
```

```python
df
```

```python
#Step 1- check the relationship of any independent feature with dependent feature.

sns.pairplot(df)
```

```python
# Step 2- seperate the input feature (x) and target feature(y).
# we have to covert the the data in values as we can not directly fir the series in ml model it should be an array.

x=df.iloc[:,:-1].values
y=df.iloc[:,1:].values
```

```python
x
```

```python
y
```

```python
#Step 3 - seperate train and test part of the data.
#dont shuffle the position of the variables.

x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.3,random_state=7)
```

```python
x_train
```

```python
x_test
```

```python
y_train
```

```python
y_test
```

```python
# Step 4- Application of model.

#read the model
from sklearn.linear_model import LinearRegression
#read the function for linear regression model.
linear_l=LinearRegression()
# .fit will fit the data inside the model.
linear_l.fit(x_train, y_train)
# the model will create the logic and will predict the values.
y_pred=linear_l.predict(x_test)
```

```python
# Step 5- Judge the accuracy of the linear regression model.

# the accuracy cant be judged as ut wont give 100% accurate results.
```

```python
# Step 5 - r2 score of the model.

linear_l.score(x_train,y_train)
```

```python
# Step 6 - finding of the error or formation of the error.

from sklearn.metrics import mean_absolute_error
mean_absolute_error(y_test,y_pred)
```

```python
# to find the cofficient.

linear_l.coef_
```

```python
# finding the error or intercept rate 

linear_l.intercept_
```

```python
import matplotlib.pyplot as plt

plt.scatter(x_test,y_test,color="red")
plt.plot(x_test,y_pred,color="blue",alpha=0.6)
```
